# Python Comments: A Complete Guide

Comments in Python are used to explain code, make it more readable, and prevent execution when testing alternative code.

In Python, there are three primary ways to think about comments and documentation:
1. **Single-line comments**
2. **Multi-line comments** (using multiple `#` symbols)
3. **Docstrings** (used for documenting modules, classes, and functions)

```python
# 1. Single-Line Comments
# Single-line comments start with the hash character (#) and extend to the end of the line.

# This is a standalone single-line comment
x = 5  # This is an inline comment explaining what 'x' represents

# 2. Multi-Line Comments
# Python doesn't have a specific syntax for multi-line comments.
# To write a multi-line comment, you simply place a hash character (#)
# at the beginning of each line.

print(f"The value of x is {x}")
```

## Docstrings (Documentation Strings)

While not technically comments, **Docstrings** are triple-quoted strings (`"""` or `'''`) used to document functions, classes, and modules. Unlike comments, which are ignored by the Python interpreter, docstrings are parsed and can be accessed at runtime using the `__doc__` attribute or the `help()` function.

```python
def calculate_area(width, height):
    """
    Calculate the area of a rectangle.

    Parameters:
    width (float): The width of the rectangle.
    height (float): The height of the rectangle.

    Returns:
    float: The calculated area.
    """
    return width * height

# Accessing the docstring
print(calculate_area.__doc__)

help(calculate_area)
```

## Using the `help()` Function

While accessing the `__doc__` attribute directly is useful, Python also provides a built-in function called `help()`.

When you pass an object (such as a function, class, or module) to `help()`, it formats and displays its docstring along with its signature in a clean, human-readable layout.

```python
# Call the help() function passing the function name (without parentheses)
help(calculate_area)
```
