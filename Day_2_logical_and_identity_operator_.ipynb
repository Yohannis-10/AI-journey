### Logical & Identity Operators in Python

#### 1. Logical Operators (`and`, `or`, `not`)
Logical operators are used to combine conditional statements. They evaluate expressions and return Boolean values (`True` or `False`).
*   `and`: Returns `True` if **both** statements are true.
*   `or`: Returns `True` if **at least one** of the statements is true.
*   `not`: Reverses the result (returns `False` if the result is true).

#### 2. Identity Operators (`is`, `is not`)
Identity operators compare the **memory location (identity)** of two objects, not just their value.
*   `is`: Returns `True` if both variables point to the **exact same object** in memory.
*   `is not`: Returns `True` if both variables point to **different objects** in memory.
*   *Crucial Note:* Use `==` to compare **values**, and use `is` to compare **identity** (commonly used when checking against `None`).

```python
# --- Practical Demonstration of Logical Operators ---
age = 25
has_license = True

# logical AND: Both conditions must be True
can_drive = (age >= 18) and has_license
print(f"Can drive? {can_drive}")

# logical OR: At least one condition must be True
has_cash = False
has_card = True
can_pay = has_cash or has_card
print(f"Can pay? {can_pay}")

# logical NOT: Negates the boolean value
is_weekend = False
is_weekday = not is_weekend
print(f"Is weekday? {is_weekday}")

print("-" * 40)

# --- Practical Demonstration of Identity Operators ---
a = [1, 2, 3]
b = [1, 2, 3]
c = a

# Value comparison (==)
print(f"Do 'a' and 'b' have the same values? {a == b}")  # True

# Identity comparison (is)
print(f"Are 'a' and 'b' the exact same object? {a is b}")  # False (different lists in memory)
print(f"Are 'a' and 'c' the exact same object? {a is c}")  # True (c points to the same list)

# Best Practice: Checking for None
user_profile = None
if user_profile is None:
    print("Profile is not set yet.")
```

```python
def validate_user_registration_robust(username, age, email):
    """
    Validates user registration data using logical operators.
    Raises ValueError if validation checks fail.
    """
    # Use 'not' to check for empty strings
    if not username:
        raise ValueError("Username cannot be empty!")

    # Use 'and' to ensure the age falls within a strict logical range
    if not (age >= 13 and age < 120):
        raise ValueError("Age must be between 13 and 119!")

    # Use 'or' to check if email is formatted correctly or if registering as a guest
    is_valid_email = "@" in email and "." in email
    is_guest = username.lower() == "guest"

    if not (is_valid_email or is_guest):
        raise ValueError("Invalid email address (only 'guest' accounts can bypass email validation)!")

    print(f"Validation Successful for user: {username}!")
    return True

# --- Test Cases demonstrating Exception Handling ---
test_cases = [
    ("alice_dev", 20, "alice@example.com"),
    ("young_coder", 10, "young@example.com"),
    ("guest", 25, ""),
    ("", 30, "test@example.com")
]

for i, (username, age, email) in enumerate(test_cases, 1):
    print(f"Running Test {i} with user '{username}':")
    try:
        validate_user_registration_robust(username, age, email)
    except ValueError as e:
        print(f"Caught expected exception: {e}")
    print("-" * 40)
```
