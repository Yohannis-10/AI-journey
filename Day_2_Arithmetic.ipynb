```python
# Python Arithmetic Operators
a = 10
b = 3

# Addition
print(f"Addition: {a} + {b} = {a + b}")

# addition
print(f"addition: {a+b}")
# Subtraction
print(f"Subtraction: {a} - {b} = {a - b}")

# Multiplication
print(f"Multiplication: {a} * {b} = {a * b}")

# Division (always returns a float)
print(f"Division: {a} / {b} = {a / b}")

# Floor Division (rounds down to the nearest integer)
print(f"Floor Division: {a} // {b} = {a // b}")

# Modulo (returns the remainder of division)
print(f"Modulo (remainder): {a} % {b} = {a % b}")

# Exponentiation (power)
print(f"Exponentiation: {a} ** {b} = {a ** b}")
```

### Operator Precedence in Python

Operator precedence determines the order in which operations are evaluated in an expression. Python follows rules similar to standard mathematics (often remembered via the acronym **PEMDAS**):

1. **P**arentheses `()`
2. **E**xponentiation `**`
3. **M**ultiplication `*`, **D**ivision `/`, Floor Division `//`, Modulo `%` (evaluated left-to-right)
4. **A**ddition `+`, **S**ubtraction `-` (evaluated left-to-right)

Let's write some code to see this precedence in action.

```python
# Demonstration of Operator Precedence

# 1. Multiplication has higher precedence than Addition
# Evaluated as: 2 + (3 * 4) = 14
expr1 = 2 + 3 * 4
print(f"2 + 3 * 4 = {expr1}")

# 2. Parentheses override default precedence
# Evaluated as: (2 + 3) * 4 = 20
expr2 = (2 + 3) * 4
print(f"(2 + 3) * 4 = {expr2}")

# 3. Exponentiation has higher precedence than multiplication
# Evaluated as: 2 * (3 ** 2) = 18
expr3 = 2 * 3 ** 2
print(f"2 * 3 ** 2 = {expr3}")
```

### Combining Arithmetic, Precedence, and Assignments

Let's look at a practical calculation that uses standard operations, respects precedence rules, and assigns the results to variables.

```python
# 1. Set up some initial value variables
x = 5
y = 2
z = 10

# 2. A complex calculation demonstrating PEMDAS
# Priority: Parentheses -> Exponentiation -> Multiplication/Division -> Addition
# Step 1: (x + y) -> (5 + 2) = 7
# Step 2: y ** 2 -> 2 ** 2 = 4
# Step 3: 7 * 4 -> 28
# Step 4: z / 2 -> 10 / 2 = 5.0
# Step 5: 28 - 5.0 -> 23.0
result = (x + y) * (y ** 2) - z / 2

print(f"Using variables: x={x}, y={y}, z={z}")
print(f"Expression: ({x} + {y}) * ({y} ** 2) - {z} / 2")
print(f"Calculated Result: {result}")
```

```python
# =====================================================
# SECTION 1: BASIC ARITHMETIC OPERATORS
# =====================================================
print("--- 1. BASIC ARITHMETIC OPERATORS ---")
num1 = 15
num2 = 4

print(f"Addition:       {num1} + {num2}  = {num1 + num2}")
print(f"Subtraction:    {num1} - {num2}  = {num1 - num2}")
print(f"Multiplication: {num1} * {num2}  = {num1 * num2}")
print(f"Division:       {num1} / {num2}  = {num1 / num2}")
print(f"Floor Division: {num1} // {num2} = {num1 // num2}")
print(f"Modulo:         {num1} % {num2}  = {num1 % num2}")
print(f"Exponentiation: {num1} ** {num2} = {num1 ** num2}")
print()

# =====================================================
# SECTION 2: OPERATOR PRECEDENCE (PEMDAS)
# =====================================================
print("--- 2. OPERATOR PRECEDENCE (PEMDAS) ---")
# Division/Multiplication happen before Addition/Subtraction
no_parentheses = 10 + 5 * 2
# Parentheses force addition first
with_parentheses = (10 + 5) * 2

print(f"Without Parentheses (10 + 5 * 2): {no_parentheses}")
print(f"With Parentheses ((10 + 5) * 2):  {with_parentheses}")
print()

# =====================================================
# SECTION 3: COMBINING THE CONCEPTS WITH ASSIGNMENT
# =====================================================
print("--- 3. COMBINED COMPLEX CALCULATION ---")
# We use multiple variables, arithmetic, and precedence
base_value = 20
multiplier = 3
power = 2
offset = 50

# Precedence Order:
# 1. (multiplier ** power) -> 3 ** 2 = 9
# 2. base_value * 9 -> 20 * 9 = 180
# 3. 180 - offset -> 180 - 50 = 130
final_result = base_value * (multiplier ** power) - offset

print(f"Variables used: base_value={base_value}, multiplier={multiplier}, power={power}, offset={offset}")
print(f"Expression: {base_value} * ({multiplier} ** {power}) - {offset}")
print(f"Final Result: {final_result}")
```

### Python Comparison Operators

Comparison operators are used to compare two values. They always return a boolean value: either `True` or `False`.

* `==` : Equal to
* `!=` : Not equal to
* `>`  : Greater than
* `<`  : Less than
* `>=` : Greater than or equal to
* `<=` : Less than or equal to

```python
# Demonstration of Comparison Operators
x = 10
y = 15

print(f"x = {x}, y = {y}\n")

print(f"Is x equal to y? (x == y):          {x == y}")
print(f"Is x not equal to y? (x != y):      {x != y}")
print(f"Is x greater than y? (x > y):       {x > y}")
print(f"Is x less than y? (x < y):          {x < y}")
print(f"Is x greater or equal to 10? (x>=10): {x >= 10}")
print(f"Is y less or equal to 12? (y<=12):    {y <= 12}")
```

### Python Logical Operators

Logical operators are used to combine conditional statements (boolean values):

* **`and`**: Returns `True` if both statements are true (e.g., `True and True` is `True`).
* **`or`**: Returns `True` if at least one of the statements is true (e.g., `True or False` is `True`).
* **`not`**: Reverses the result, returning `False` if the result is true, and vice-versa (e.g., `not True` is `False`).

```python
# Demonstration of Logical Operators
a = True
b = False

print(f"a = {a}, b = {b}\n")

# logical AND
print(f"a and b: {a and b}  (Both must be True)")

# logical OR
print(f"a or b:  {a or b}   (At least one must be True)")

# logical NOT
print(f"not a:   {not a}  (Opposite of a)")
print(f"not b:   {not b}   (Opposite of b)")

print("\n--- Real-world Precedence Example ---")
age = 22
has_ticket = True
is_banned = False

# Can enter if they are older than 18 AND have a ticket, AND are NOT banned
can_enter = (age >= 18) and has_ticket and (not is_banned)
print(f"Can the person enter? {can_enter}")
```
