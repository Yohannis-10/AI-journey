# Pythonic Conditional Statements (If-Else Patterns)

This notebook demonstrates how to write practical, clean, and highly readable conditional logic in Python. We will cover:
1. Standard conditional structure
2. The Ternary Operator (One-liners)
3. Avoid Deep Nesting using 'Guard Clauses'
4. Using Dictionaries instead of complex `elif` chains
5. Structural Pattern Matching (`match-case` introduced in Python 3.10)

### 1. Standard If-Else with Multiple Conditions
Use explicit conditions, parentheses are not required. Ensure readability by keeping expressions simple.

```python
age = 20
has_permit = True

# Clean, readable multiple-condition checks using 'and' / 'or'
if age >= 18 and has_permit:
    print("You are allowed to drive.")
else:
    print("You cannot drive.")
```

### 2. The Ternary Operator (One-liner)
When assigning a value based on a condition, avoid writing 4 lines of code. Use a ternary operator instead.

```python
score = 85

# Pythonic ternary structure: [value_if_true] if [condition] else [value_if_false]
status = "Passed" if score >= 50 else "Failed"
print(f"Result: {status}")
```

### 3. Guard Clauses (The "Flat is Better Than Nested" rule)
Deeply nested `if` statements are hard to read. Return early using "guard clauses" to keep your code linear and flat.

```python
# Unclean (Nested Structure):
def register_user_nested(user):
    if user.get("username"):
        if user.get("age") >= 18:
            return "Registration successful"
        else:
            return "Error: User must be 18 or older"
    else:
        return "Error: Missing username"

# Clean (Flat Structure using Guard Clauses):
def register_user_clean(user):
    if not user.get("username"):
        return "Error: Missing username"

    if user.get("age") < 18:
        return "Error: User must be 18 or older"

    return "Registration successful"

test_user = {"username": "Alice", "age": 20}
print(register_user_clean(test_user))
```

### 4. Replacing Complex Elif Chains with Dictionaries
If you find yourself writing too many `elif` statements mapping a key to a function or value, a dictionary mapping is often more elegant.

```python
def get_role_permissions(role):
    # Using a dictionary as a clean dispatcher
    permissions = {
        "admin": "All access",
        "editor": "Edit content",
        "viewer": "Read-only access"
    }
    # .get() handles missing roles elegantly using a default value
    return permissions.get(role, "Guest access")

print(f"Admin: {get_role_permissions('admin')}")
print(f"Anonymous: {get_role_permissions('anonymous')}")
```

### 5. Structural Pattern Matching (Python 3.10+)
For modern Python, the `match` statement provides a robust alternative to `if-elif-else` patterns.

```python
http_status = 404

match http_status:
    case 200:
        print("Success!")
    case 400 | 404:
        print("Client Error (Bad Request or Not Found)")
    case 500:
        print("Server Error")
    case _: # Wildcard acts as the default 'else'
        print("Unknown Status Code")
```
