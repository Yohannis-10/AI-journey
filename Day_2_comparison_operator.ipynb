# Python Comparison Operators in Practice

Comparison operators are used to compare two values. They return a boolean value: either `True` or `False`. Here are the standard comparison operators in Python:

* `==` (Equal to)
* `!=` (Not equal to)
* `>` (Greater than)
* `<` (Less than)
* `>=` (Greater than or equal to)
* `<=` (Less than or equal to)

Let's see them in a practical real-world scenario, such as an e-commerce checkout check.

```python
# Practical Scenario: E-commerce Order Verification

cart_total = 120.50
minimum_for_free_shipping = 100.00
user_age = 17
age_restriction_limit = 18
coupon_code = "DISCOUNT20"
applied_coupon = "DISCOUNT20"

# 1. Greater than or equal to (>=) to check for free shipping
eligible_for_free_shipping = cart_total >= minimum_for_free_shipping
print(f"Is eligible for free shipping? {eligible_for_free_shipping}")
```

```python
# 2. Less than (<) to check age limit
is_underage = user_age < age_restriction_limit
print(f"Is the user underage? {is_underage}")

# 3. Equal to (==) to verify coupon code validity
is_coupon_valid = coupon_code == applied_coupon
print(f"Is the coupon code valid? {is_coupon_valid}")

# 4. Not equal to (!=) to check if any extra fees apply
shipping_method = "Express"
extra_fee_applies = shipping_method != "Standard"
print(f"Does an extra shipping fee apply? {extra_fee_applies}")
```

## Combining Comparison Operators with Logical Operators

To build more complex decision-making rules, you can combine comparison operators using **Logical Operators**:

* **`and`**: Returns `True` if **both** comparisons are true.
* **`or`**: Returns `True` if **at least one** comparison is true.
* **`not`**: Reverses the boolean result (turns `True` into `False` and vice versa).

```python
# Let's check complex conditions using logical operators

# Scenario A: Can the customer buy a restricted item with free shipping?
# Both conditions must be met: NOT underage AND eligible for free shipping
can_buy_restricted_with_free_shipping = (not is_underage) and eligible_for_free_shipping
print(f"Can buy restricted item with free shipping? {can_buy_restricted_with_free_shipping}")

# Scenario B: Can the user get a discount?
# At least one condition must be met: Have a valid coupon OR spend more than $150
large_spender_limit = 150.00
is_large_spender = cart_total > large_spender_limit

eligible_for_discount = is_coupon_valid or is_large_spender
print(f"Is eligible for any discount? {eligible_for_discount}")
```

## Creating an Eligibility Verification Function

Let's encapsulate this logic into a reusable Python function. This function will take the customer's age and cart total as arguments and return whether they are allowed to purchase (must be 18 or older) and if they qualify for free shipping (must spend $100 or more).

```python
def check_checkout_eligibility(age, total):
    """
    Checks user eligibility for age-restricted purchases and free shipping.
    """
    # Comparison checks
    is_adult = age >= 18
    has_free_shipping = total >= 100.00

    # Combining eligibility status
    can_checkout = is_adult

    print(f"Customer Age: {age} | Cart Total: ${total:.2f}")
    print(f"- Can complete purchase (18+)? {can_checkout}")
    print(f"- Qualifies for free shipping ($100+)? {has_free_shipping}")
    print("-" * 40)

    return {
        "can_purchase": can_checkout,
        "free_shipping": has_free_shipping
    }

# Test the function with different customer scenarios
print("Scenario 1: Underage buyer with small cart")
check_checkout_eligibility(age=16, total=45.00)

print("Scenario 2: Adult buyer with large cart")
check_checkout_eligibility(age=25, total=120.50)
```
