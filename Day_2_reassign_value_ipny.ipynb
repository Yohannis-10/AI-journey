# Understanding Variable Reassignment in Python

In Python, variables do not work like "boxes" that hold values. Instead, they work like **labels** or **references** pointing to objects in memory.

Let's explore variable reassignment, how memory references change, and the crucial difference between reassigning a variable and mutating an object.

## 1. Basic Reassignment (Immutable Objects)

When you reassign a variable pointing to an immutable object (like an integer, string, or tuple), Python creates a new object in memory and updates the variable to point to this new object. The original object is eventually cleaned up by Python's garbage collector if nothing else references it.

```python
# Define a variable pointing to an integer
x = 10
print(f"Initial x: {x}, memory address (id): {id(x)}")

# Reassign x to a new integer
x = 20
print(f"Reassigned x: {x}, memory address (id): {id(x)}")

# Note how the 'id' (memory location) changed. x is now pointing to a completely different object.
```

## 2. Reassignment vs. In-Place Mutation (Mutable Objects)

It is vital to distinguish between **reassigning** a variable and **mutating** the object it points to. This distinction is clearest when working with mutable objects like lists or dictionaries.

```python
# --- Scenario A: Mutation (In-place modification) ---
a = [1, 2, 3]
b = a  # b references the SAME list object in memory

print("Before mutation:")
print(f"a: {a} (id: {id(a)})")
print(f"b: {b} (id: {id(b)})")

a.append(4)  # Mutating the list object in-place

print("\nAfter mutating a:")
print(f"a: {a} (id: {id(a)})")
print(f"b: {b} (id: {id(b)}) - b changed too because it points to the same object!")

# --- Scenario B: Reassignment ---
print("\n" + "="*40 + "\n")

a = [1, 2, 3]
b = a

print("Before reassignment:")
print(f"a: {a} (id: {id(a)})")
print(f"b: {b} (id: {id(b)})")

a = [1, 2, 3, 4]  # Reassigning 'a' to a BRAND NEW list object

print("\nAfter reassigning a:")
print(f"a: {a} (id: {id(a)}) - points to a new ID")
print(f"b: {b} (id: {id(b)}) - remains unchanged, pointing to the original list")
```

## 3. Variable Reassignment with Custom Classes

Just like lists and dictionaries, instances of custom Python classes are mutable objects. When you assign a class instance to multiple variables, they all point to the same object in memory.

```python
class Player:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def __repr__(self):
        return f"Player(name='{self.name}', score={self.score})"

# Create a new instance of Player
player_one = Player("Jorgoo", 100)

# Create another reference to the same object
player_two = player_one

print("Initial State:")
print(f"player_one: {player_one} (id: {id(player_one)})")
print(f"player_two: {player_two} (id: {id(player_two)})")

# --- Mutation: Modifying an attribute in-place ---
player_one.score = 150
print("\nAfter mutating player_one.score:")
print(f"player_one: {player_one}")
print(f"player_two: {player_two} (also updated because it points to the same object)")

# --- Reassignment: Pointing to a new object ---
player_one = Player("Bob", 200)  # Reassigning player_one to a new instance
print("\nAfter reassigning player_one:")
print(f"player_one: {player_one} (id: {id(player_one)}) - points to Bob")
print(f"player_two: {player_two} (id: {id(player_two)}) - still points to Alice")
```
